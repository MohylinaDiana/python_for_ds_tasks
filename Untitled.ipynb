```python
import datetime
import requests
import json
import os

import matplotlib.pyplot as plt
from dotenv import load_dotenv
import pandas as pd
import sqlalchemy as sa
from sqlalchemy import create_engine, text, MetaData, Table
from sqlalchemy.orm import sessionmaker
```

```python
def create_connection():
    """
    Створює підключення через SQLAlchemy
    """
    # Завантажуємо змінні середовища
    load_dotenv()

    # Отримуємо параметри з environment variables
    host = os.getenv('DB_HOST', 'localhost')
    port = os.getenv('DB_PORT', '3306')
    user = os.getenv('DB_USER')
    password = os.getenv('DB_PASSWORD')
    database = os.getenv('DB_NAME')

    if not all([user, password, database]):
        raise ValueError("Не всі параметри БД задані в .env файлі!")

    # Створюємо connection string
    connection_string = f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}"

    # Створюємо engine з connection pooling
    engine = create_engine(
        connection_string,
        pool_size=2,           # Розмір пулу підключень
        max_overflow=20,        # Максимальна кількість додаткових підключень
        pool_pre_ping=True,     # Перевірка підключення перед використанням
        echo=False              # Логування SQL запитів (True для debug)
    )

    # Тестуємо підключення
    try:
        with engine.connect() as conn:
            result = conn.execute(text("SELECT 1"))
            result.fetchone()

        print("✅ Підключення до БД успішне!")
        print(f"🔗 {user}@{host}:{port}/{database}")
        print(f"⚡ Engine: {engine}")

        return engine

    except Exception as e:
        print(f"❌ Помилка підключення: {e}")
        return None

# Створюємо підключення
engine = create_connection()
```

```python
query3 = text("""
WITH PR AS (SELECT p.productName, p.productLine, SUM(od.quantityOrdered * od.priceEach) as rev FROM products p JOIN orderdetails od ON p.productCode = od.productCode GROUP BY p.productCode),
TR AS (SELECT SUM(rev) as total FROM PR)
SELECT productName, productLine, rev, RANK() OVER (ORDER BY rev DESC) as rnk, (rev / (SELECT total FROM TR)) * 100 as pct, (rev / AVG(rev) OVER (PARTITION BY productLine) - 1) * 100 as diff FROM PR ORDER BY pct DESC
""")
df3 = pd.read_sql(query3, engine)

print(f"Top 1: {df3.iloc[0]['productName']} ({df3.iloc[0]['pct']:.2f}%)")

df3.head(10).plot(kind='barh', x='productName', y='rev', figsize=(10, 5))
plt.show()

line_rev = df3.groupby('productLine')['rev'].sum().sort_values(ascending=False)
line_rev.plot(kind='pie', autopct='%1.1f%%', figsize=(7, 7))
plt.show()

df3['cum_pct'] = df3['pct'].cumsum()
pareto = len(df3[df3['cum_pct'] <= 80]) + 1

print(f"Ratio 1st/10th: {df3.iloc[0]['rev'] / df3.iloc[9]['rev']:.2f}")
print(f"Top 2 Lines Pct: {(line_rev.iloc[:2].sum() / line_rev.sum()) * 100:.2f}%")
print(f"Pareto Count: {pareto} products")
```
