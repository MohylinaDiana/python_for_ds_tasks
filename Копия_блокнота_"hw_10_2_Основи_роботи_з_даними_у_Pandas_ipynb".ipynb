9. Врахуовуючи висновок в попередньому завданні, дослідимо як продаються в цих містах різні категорії товарів. Виведіть відсотковий розпоіділ кількості інвойсів по категоріям товарів `Product line` в містах `Yangon` та `Mandalay`.

```python
yangon_data = df[df['City'] == 'Yangon']['Product line'].value_counts(normalize=True)
mandalay_data = df[df['City'] == 'Mandalay']['Product line'].value_counts(normalize=True)
result = pd.concat([yangon_data, mandalay_data], axis=1, keys=['Yangon', 'Mandalay'])

print(result)
```
